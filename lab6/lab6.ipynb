### Pеалізувати усі фрактали з підрозділів 1.4 та 1.5 даної лабораторної роботи.

```python
def generate_fractal(set_of_values, set_of_probabilities):
    
    n_points = 50000
    x, y = 0, 0
    x_new, y_new = [], []

    if set_of_probabilities == 0:
        p = 1 / len(set_of_values)
        set_of_probabilities = [p] * len(set_of_values)


    for _ in range(n_points):
        
        a, b, c, d, e, f = random.choices(set_of_values, weights = set_of_probabilities, k=1)[0]

        x_n = a * x + b * y + e
        y_n = c * x + d * y + f

        x, y = x_n, y_n

        x_new.append(x)
        y_new.append(y)

    return np.array(x_new), np.array(y_new)
```

```python
def generate_fractal_polar_coordinates(set_of_values, set_of_probabilities):
    
    n_points = 50000
    x, y = 0, 0
    x_new, y_new = [], []

    if set_of_probabilities == 0:
        p = 1 / len(set_of_values)
        set_of_probabilities = [p] * len(set_of_values)


    for _ in range(n_points):
        
        r, s, teta, fi, e, f = random.choices(set_of_values, weights = set_of_probabilities, k=1)[0]

        x_n = r * np.cos(teta)* x - s *np.sin(fi)* y + e
        y_n = r *np.sin(teta)* x + s * np.cos(fi)*y + f

        x, y = x_n, y_n

        x_new.append(x)
        y_new.append(y)

    return x_new, y_new
```

### Модифікувати програмне забезпечення динамічною (по ітераціях) візуалізацією кожного фракталу.

```python
def fractal_animation(x, y, color, output_gif):
    
    n = 200
    total = 100000
    frames = total // n

    img_h, img_w = 600, 600
    
    x_min, x_max = np.min(x), np.max(x)
    y_min, y_max = np.min(y), np.max(y)

    x = 2 * (x - x_min) / (x_max - x_min) - 1
    y = 2 * (y - y_min) / (y_max - y_min) - 1
    
    img = np.ones((img_h, img_w, 3), dtype=np.uint8) * 255
    
    images = []

    for k in range(frames):
        
        start = k * n
        end = start + n
        x_frame = x[start:end]
        y_frame = y[start:end]

        for xi, yi in zip(x_frame, y_frame):
            
            px = int((xi + 1) * img_w / 2)
            py = int((1 - yi) * img_h / 2) 

            if 0 <= px < img_w and 0 <= py < img_h:
                img[py, px] = color
            
        images.append(cv2.cvtColor(img.copy(), cv2.COLOR_BGR2RGB))
                    
        cv2.imshow("Fractal", img)
        if cv2.waitKey(50) & 0xFF == ord('q'):
            break

    cv2.destroyAllWindows()
    
    imageio.mimsave(output_gif, images, fps=12)
```
