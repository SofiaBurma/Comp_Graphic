```python
import matplotlib.pyplot as plt
import math
```


- N — кількість самоподібних частин, на які фрактал розбивається на кожному рівні.
- r — коефіцієнт масштабу, який показує, у скільки разів зменшується кожна з частин на кожному рівні.

```python
numbers = ['0', '1', '2', '4', '5', '6', '8', '9']

N = len(numbers)  
r = 1 / 10              
dimension = math.log(N) / math.log(1 / r)

print(f"Фрактальна розмірність = {dimension:.10f}")

def generate_fractal(level, max_level, prefix=''):
    
    if level == max_level:
        value = float('0.' + prefix)
        return [value]
    
    result = []
    
    for n in numbers:
        result += generate_fractal(level + 1, max_level, prefix + n)
    return result

points = generate_fractal(0, 4)

plt.figure(figsize=(12, 2))
plt.scatter(points, [0]*len(points), s=1, color='green')
plt.title(f'Фрактал на [0, 1] без цифр 3 та 7')
plt.grid(True)
plt.show()
```
